```python
from numpy import loadtxt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Normalization
```

```python
# load the dataset
dataset = loadtxt('pima-indians-diabetes.csv', delimiter=',')

# split into input (X) and output (y) variables
X = dataset[:, 0:8]
y = dataset[:, 8]
```

```python
# normalize the input data
normalizer = Normalization(axis=-1)
normalizer.adapt(X)
X = normalizer(X)
```

```python
# define the keras model
model = Sequential()
model.add(Dense(64, input_shape=(8,), activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
# compile the keras model
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# fit the keras model using the original dataset as validation
history = model.fit(X, y, epochs=120, batch_size=16, verbose=0)
```

```python
# evaluate the keras model
_, accuracy = model.evaluate(X, y)
print('Accuracy: %.2f' % (accuracy * 100))
```

```python
# denormalize the input data
mu = normalizer.mean.numpy() # get the mean of the normalizer
sigma = (normalizer.variance.numpy()) ** 0.5  # get standard deviation of the normalizer
X_denormalized = (X.numpy() * sigma) + mu  # reverse normalization
```

```python
# make class predictions with the model  
predictions = (model.predict(X) > 0.5).astype(int)
```

```python
# summarize the first 5 cases
for i in range(5):
    print('%s => %d (expected %d)' % (X_denormalized[i].tolist(), predictions[i], y[i]))  
```
